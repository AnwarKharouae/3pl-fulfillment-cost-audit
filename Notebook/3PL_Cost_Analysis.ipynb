```python
start_date = pd.Timestamp('2022-03-31')
end_date   = pd.Timestamp('2022-06-29')
```

```python
#outbound
filt=Merchant['Status'].isin(outbound)
Shiprocket_outbound  = (Merchant[filt]['Qty'] * 7 ).sum()
INCREFF_outbound     = (Merchant[filt]['Qty'] * 11 ).sum()
```

```python
#canceled_stock
filt_canceled= Merchant['Status']=='Cancelled'
temp= Merchant[filt]['Qty'].replace(0,1)
print(temp)
canceled_stock_ship     = (temp[filt]*3 *25).sum()
canceled_stocl_incref   = (temp[filt]*90 *0.15).sum()
```

```python
Merchant['Status']=='Cancelled'
```

```python
#outbound stock
filt= Merchant['Status'].isin(outbound)
days= (Merchant[filt]['Date']- start_date ).dt.days
days= days.replace(0,1)

outbound_stock_ship = (Merchant[filt]['Qty'] * (days/30) * 25).sum()
outbound_stock_ince =  (Merchant[filt]['Qty'] * days * 0.15 ).sum()
```

```python
#retour stock
filt= Merchant['Status'].isin(retour)
days= (end_date-Merchant[filt]['Date'] ).dt.days
days= days.replace(0,1)
retour_stock_ship = (Merchant[filt]['Qty'] * (days/30) * 25).sum()
retour_stock_ince =  (Merchant[filt]['Qty'] * days * 0.15 ).sum()
```

```python
#retour cost
filt= Merchant['Status'].isin(retour)
retour_Cost_ship = (Merchant[filt]['Qty'] *  6  ).sum()
retour_Cost_ince = (Merchant[filt]['Qty'] *   15.5   ).sum()
```

```python
Amazon_merchant_Cost = pd.DataFrame({
    'Cost_element' :['Outbound','Outbound_Storage','Canceled_Storage','Retour_Stock','Retour_Cost'],
    'SHIPROCKET'   :[Shiprocket_outbound,outbound_stock_ship,canceled_stock_ship,retour_stock_ship,retour_Cost_ship],
    'INCREFF'      :[INCREFF_outbound,outbound_stock_ince,canceled_stocl_incref,retour_stock_ince,retour_Cost_ince],
})
```
